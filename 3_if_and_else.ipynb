### ***If ,elif, else rules.***  
• if (Only once per code block(Mandtory))  
• elif(as many time as per code block(optional))  
• else(Only once per code black(optional))  

```python
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

### ***Nested if & else***

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b:
    if a > c:
        print("A is the largest")
    else:
        print("C is the largest")
else:
    if b > c:
        print("B is the largest")
    else:
        print("C is the largest")
```

### ***short hand if & else***

```python
a = 10
b = 20

print("A is bigger") if a > b else print("B is bigger")
```
