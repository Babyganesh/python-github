### 3. Explain type casting with two examples.  
    Type casting means converting one data type into another data type.


```python
#  int() → converts to integer
a = "100"
b = int(a)
print("int():", b, type(b))   

#  tuple() is a function used to convert a value into a tuple. 
A=(10,20,30,40,50)
print(A,type(A))
print(A[0])
print(A[4])
```

 ### 4.What is the difference between a list and a tuple?  
  
   ***List***  
   Written using []  
   Mutable – can be changed  
   Example: [10, 20, 30]  
   Ordered and changeable.  
   ***Tuple***   
   Written using ()  
   Immutable – cannot be changed  
   Example: (10, 20, 30)  
   Ordered and unchangeable.

	


### 5. What is the difference between a set and a dictionary?
***Set***  
Written using {}  
Stores only values  
unordered and unindexed. No duplicate  
Example: {0, 20, 30}  
Used to store unique items  
***dictionary***  
Written using {key: value}  
Stores key-value pairs  
Unordered, changeable and indexed. No duplicate keys.  
Example: {"name": "John", "age": 20}  
Used to store data with labels/keys
