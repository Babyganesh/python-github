**1.Define a class Car with attributes brand and model. Create an object of the class and display its details.**

**3.Create a class Rectangle with attributes length and width. Add methods to calculate area and perimeter.**

```python
class Rectangle:

    def area(self, length, width):
        return length * width

    def perimeter(self, length, width):
        return 2 * (length + width)


obj = Rectangle()

print("Area:", obj.area(4, 5))
print("Perimeter:", obj.perimeter(4, 5))
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return f"Area: {self.length * self.width}"

    def perimeter(self):
        return f"Perimeter: {2 * (self.length + self.width)}"


obj = Rectangle(4, 5)

print(obj.area())
print(obj.perimeter())
```

**5.Define a class Circle with a class variable pi = 3.14. Add a method to calculate the area.**

```python
class Circle:
    pi = 3.14   # Class variable

    def __init__(self, radius):
        #instance variable
        self.radius = radius

    def calculate_area(self):#here self is main object (area=pi*r*r)
        return Circle.pi * self.radius * self.radius 


obj = Circle(4)

print("Area:", obj.calculate_area())         
```

**6.Create a class BankAccount with attributes account_number, name, and balance. Add methods to deposit, withdraw, and display balance.**

```python
class BankAccount:
    def __init__(self,account_number,name,balance):
        self.account_number=account_number
        self.name=name
        self.balance=balance

    def deposit(self,amount):
        if amount>0:
            self.balance+=amount
            print(f"successfully deposited ${amount}. New balance: ${self.balance}")
        else:
            print("deposit amount must be greater than zero.")

    def withdraw(self,amount):
        if amount>self.balance:
            print(f"transaction declined: Insufficient balance. Available: ${self.balance}")    
        else:
            self.balance -= amount
            print(f"Successfully withdrew ${amount}. Remaining balance: ${self.balance}")    
            
    def display_balance(self):
        print("Account Number:", self.account_number)
        print("Name:", self.name)
        print("Balance:", self.balance)

obj=BankAccount(8984685983445,"baby",400)

obj.deposit(100)
obj.withdraw(500)
obj.display_balance()
```

**7.Write a class Employee with attributes emp_id, name, and salary. Add a method to give a 10% bonus to employees.**



```python
class Employee:
    def __init__(self,emp_id,name,salary):
        self.emp_id=emp_id
        self.name=name
        self.salary=salary

    def gives_bonus(self):
        bonus=self.salary*10/100   #here 10%=10/100
        self.salary+=bonus
        return bonus
    

obj=Employee(1234,"baby",50000)
print("Employee_id: ",obj.emp_id)
print("name: ",obj.name)
print("salary: ",obj.salary)
print()
print(f"initial salary :{obj.salary}")
obj.gives_bonus()
print(f"Salary after 10% bonus salary :{obj.salary}")
```

**8.Implement a class Library with methods to add_book(title), remove_book(title), and display_books().**

```python
from pprint import pprint

class Library:
    def __init__(self):
        # Initialize an empty list to store book titles
        self.books = []

    def add_book(self,title):
        self.books.append(title)
        print(f"{title}, added successfully.")

    def remove_book(self,title):
        if title in self.books:
            self.books.remove(title)
            print(f"{title},remove successfully")
        else:
            print(f"{title},not found")

    def display_book(self):
        print("books in library")
        pprint(self.books)   #to avoid [ books]-->csn use for loop

        #for each in self.books:
        #     print(each)

obj=Library()

obj.add_book("Python")
obj.add_book("Java")
obj.remove_book("Java")

obj.display_book()

```

**9.Create a class Calculator with methods add, subtract, multiply, and divide. Demonstrate with objects.**

```python
class Calculator:
      def add(self, a, b):
        return a + b

      def subtract(self, a, b):
        return a - b

      def multiply(self, a, b):
        return a * b

      def divide(self, a, b):
        return a / b


# Create object
obj = Calculator()

print("Addition:", obj.add(20, 5))
print("Subtraction:", obj.subtract(10, 5))
print("Multiplication:", obj.multiply(3, 5))
print("Division:", obj.divide(30, 5))
```
