#### **2.Explain the role of the try, except, else, and finally blocks.**  
 * In Python, try, except, and finally are blocks used for exception handling,which prevents your program when an error occurs 

 * try:  contains the code that might cause an error (exception). Python will attempt to run this code first.  

 * except: it catches and handles the error if one occurs inside the try block. we can specify particular errors to catch, like ZeroDivisionError or ValueError.

 * finally: This is always runs, whether an error occurred . It is  used for cleanup actions

#### **8.Can we use multiple except blocks in Python? Give an example.**
* Yes, we can use multiple except blocks with a single try block . This allows you to handle different types of errors in unique ways .When an error occurs inside the try block, Python checks the except blocks from top to bottom and executes only the first matching one.

```python
try:
    number = int(input("Enter a number to divide 100 by "))    
    result = 100 / number    
    print(f" The result is {result}")

except ValueError:
     print(" You must enter a valid whole number")

except ZeroDivisionError:
      print(" You cannot divide a number by zero")

except Exception as e:
    # any other unexpected error
    print(f" unexpected error occurred: {e}")
```

#### **9. What happens if you use a finally block without an except block?**  
* if you use a finally block without an except block,the code inside the finally block will still execute, but any exception thrown in the try block will not be handled.
* The try block runs.
* If an exception occurs, execution of the try block stops immediately.
* Control shifts directly to the finally block, which executes completely.
* Once the finally block finishes, the original exception is raised in the console.

#### **Task 1: List Index and Input Error** 
Create a program that:  
• Gets a list of numbers from the user.  
• Asks the user to enter an index.  
• Use nested try-except blocks to handle:  
• ValueError  
• IndexError  

```python
numbers = [10, 20, 30, 40, 50]
print("Your list of numbers is:", numbers)


try:
    # Ask the user to enter an index
    user_input = input("Enter an index to look up: ")
    
    try:
        #  convert the input into an integer
        index = int(user_input)  #  raise ValueError
        
        #  access the list item at that index
        result = numbers[index]  #  raise IndexError
        
        print(f"The number at index {index} is {result}.")
        
    except IndexError:
        #   index is out of range
        print("Error: That index does not exist in the list!")
        
except ValueError:
    #    entered  instead of a number
    print("Error: You must enter a valid whole number!")
```

```python
try:
    n = int(input("enter int number: "))  # LIKE n=5
    mylist = [] # n is convert to list [0......5]
    print(list(mylist))

    for i in range(n): # range(5)
        num = int(input("Enter number: ")) # enter 5 times like 10,20,30,40,50    
        mylist.append(num) #adds an item to the end of the list 
                            # [10,20,30,40,50]
        print(list(mylist))

    try:
        index = int(input("Enter an index: ")) # like 3  
        print(f"Value at index {index}: {mylist[index]}") #value=mylist[3]=40

    except ValueError: #entered  instead of a number
        print("Enter a valid index.")

except IndexError: #entered  out of range index 
    print("Index is out of range.")
```
