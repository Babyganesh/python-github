## **while loop**

```python
# **print the bellow pattern**   
""" *****
    *****
    *****
    *****
    ***** """

a=1
while a<=5: # row
    b=1
    while b<=5: #column
        print("*",end="")
        b += 1
    a += 1
    print()
```

```python
# while loop(FROM USER)
row=int(input("enter a row number "))
column=int(input("enter a column number"))
symbol=input("enter the symbol")

a=1
while a<=row: # row
    b=1
    while b<=column: #column
        print(symbol,end="")
        b += 1
    a += 1
    print()
```

```python
# while loop
# *,**,**,***,****
row = int(input("Enter a row number: "))
symbol = input("Enter the symbol: ")

a = 1
while a <= row:
    b = 1
    while b <= a:
        print(symbol, end="")
        b += 1

    print()
    a += 1    
```

### **for loop pattern**

```python
""" print
    ****  
    ****
    ****
    **** """

row = int(input("Enter the number of rows: "))
column= int(input("Enter the number of column: "))
symbol = input("Enter the symbol: ")

for i in range(row):
    for j in range(column):
        print(symbol, end="")
    print()
```

```python
#  print the symbol:*,**,**,***,**** 

n = int(input("Enter the number of rows: "))
symbol = input("Enter the symbol: ")

for i in range(n):
    for j in range(0,i + 1):
        print(symbol, end="")
    print()
```

```python
for x in range(1,6):    
        print("^"*x)
       
```

```python
#print the below pattern 
""" &&&&&
    &&&& 
    &&&
    &&
    & """

n = int(input("Enter the number of rows: "))
# symbol = input("Enter the symbol: ")

for i in range(n):
    for j in range(n-i):
        print("&", end="")
    print()
```
