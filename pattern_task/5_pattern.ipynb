## **while loop**

```python
# **print the bellow pattern**   
""" *****
    *****
    *****
    *****
    ***** """

a=1
while a<=5: # row
    b=1
    while b<=5: #column
        print("*",end="")
        b += 1
    a += 1
    print()
```

```python
# while loop
# *,**,**,***,****
row = int(input("Enter a row number: "))
symbol = input("Enter the symbol: ")

a = 1
while a <= row:
    b = 1
    while b <= a:
        print(symbol, end="")
        b += 1

    print()
    a += 1    
```

```python
#print the below pattern vwhile loop
""" &&&&&
    &&&& 
    &&&
    &&
    & """

n=int(input("enter the number"))
symbol = input("Enter the symbol: ")

i = 1

while i <= n:
    j = n

    while j >= i:
        print(symbol, end="")
        j -= 1

    print()
    i += 1
```

### **for loop pattern**

```python
""" print
    ****  
    ****
    ****
    **** """

row = int(input("Enter the number of rows: "))
column= int(input("Enter the number of column: "))
symbol = input("Enter the symbol: ")

for i in range(row):
    for j in range(column):
        print(symbol, end="")
    print()
```

```python
#  print the symbol:*,**,**,***,**** 

n = int(input("Enter the number of rows: "))
symbol = input("Enter the symbol: ")

for i in range(n):
    for j in range(0,i + 1):
        print(symbol, end="")
    print()
```

```python
# Print the blow pattern (10/9/2026)
"""  #
    ##
   ###
  ####
 ##### """

n = int(input("Enter the number of rows: "))

for i in range(1, n):
    for j in range(n - i):
        print(" ", end="")
    for j in range(i):
        print("#", end="")
    print()
```

```python
# Print the blow pattern
""" ------
     -----
      ----
       ---
        --
         -
    """

n = int(input("Enter the number of rows: "))

for i in range(n):
    for j in range(i+1):
        print(" ", end="")
    for j in range(i,n):
        print("-", end="")
    print()
```
