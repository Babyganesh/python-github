```python
import re

def information(text):
    emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)
    phones = re.findall(r'\d{10}', text)
    urls = re.findall(r'\shttps?:\W+\w+\W\w+.\w+', text)

    # Return Results
    return emails, phones, urls

# opean and read the data in file
with open("C:\\Users\\Admin\\Desktop\\filehandling\\contacts.txt", "r") as file:
    data = file.read()

# Call the Function
emails, phones, urls = information(data)

# Display Results
print("Emails:", emails)
print("Phone Numbers:", phones)
print("URLs:", urls)
```
