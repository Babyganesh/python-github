```python
# game program
score=100
secret_number=50
guess =int(input("enter the guess:" ))
if guess==secret_number:
    print("Correct")
    score=score+20
    print("your score ;",score)

elif guess>secret_number:
    print("too high")
    score=score-10
    print("your score ;",score)   

else:  
    print(" too low")
    score=score-10
    print("your score ;",score)
```

```python

n = int(input("Enter the number of rows: "))

for i in range(n):
    for j in range(0,n+1):
        print("*", end="")
    print()
```

```python
#for loop parttern
n = int(input("Enter the number of rows: "))
symbol = input("Enter the symbol: ")

for i in range(n):
    for j in range(i + 1):
        print(symbol, end="")
    print()
```
