```python
# game program
score=100
secret_number=50
guess =int(input("enter the guess:" ))
if guess==secret_number:
    print("Correct")
    score=score+20
    print("your score ;",score)

elif guess>secret_number:
    print("too high")
    score=score-10
    print("your score ;",score)   

else:  
    print(" too low")
    score=score-10
    print("your score ;",score)
```
