***PYTHON OPERATORS***

```python
# Arithmetic operators(+,-, *, / ,**,//(Quotient),%(remainder))

a = int(input("Enter 1st num= "))
b = int(input("Enter 2nd num= "))
print("Enter 1st num= ",a)
print("Enter 2nd num= ",b)
print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Power =", a ** b)
print("Quotient =", a // b)
print("Remainder =", a % b)
```

```python
# Assignment operators  (=, +=,-=, /=, *= , **=, //=,%=)
a=45
print("fist value =",a)
a+=50
print("after a+=50:",a)
a-=50
print("after a-=50:",a)
a/=50
print("after a/=50:",a)
a*=50
print("after a*=50:",a)
a**=50
print("after a**=50:",a)
a//=50
print("after a//=50:",a)
a%=50
print("after a%=50:",a)
```

```python
# Comparison operators (==, !=, >, <, <=, >=) (like true or false)
a=10
b=40
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a<=b)
print(a>=b)
```

```python
# Logical operators (and, or, not)
a=90
b=60    # ex :int
c=98
print("-----int----")
print(a>b and b<=c) # t and t=t
print(b!=a or a==c) # t or f=t
print(not(c>=b))    # (not (t))=false

print("-----str----")
a="BABY"
b="baby"
print(a==b and a!=b) # f and t=f
print(a!=b and a!=b) # t and t=t
```

***PYTHON COLLECTION***

 ***list*** is a collection which is ordered and changeable. Allows duplicate,members.  
Denoted by[].  
***Tuple*** is a collection which is ordered and unchangeable.  
 Allows duplicate members. Denoted by().  
***Set*** is a collection which is unordered and unindexed.  
 No duplicate members. Denoted by{}  
***Dictionary*** is a collection which is unordered, changeable and indexed.   
No duplicate keys. Denoted by {key:value}  

```python
# List[]
# in python valuse starts from 0's
# a=[34,56,89,90,89,]=[0,1,2,3,4,5]

mylist=["bus",0,"car",0,"bike",45.7,"train",34,"flight","truck",90.98,"train"]
print(mylist,type(mylist))
print(mylist[0])
print(mylist[1])
print(mylist[2])
print(mylist[3])
mylist[4]=56 # change index value
print(mylist)

print(mylist.append("baby")) #add at end
print(mylist)
```

```python
# Tuple[]

mytuple=[]
print(mytuple,type(mytuple))

mytuple=("bus",0,"car",0,"bike",45.7,"train",34,"flight","truck",90.98,"train")
print(mytuple,type(mytuple))
print(mytuple[0])
print(mytuple[1])
print(mytuple[2])
print(mytuple[3])
mytuple[4]=56 # cannot change index value
print(mytuple)
```

```python
# set{} --->The o/p will be a dict bcz it depends on the values given.
myset={}
print(myset,type(myset)) 
myset={23.4,False,90,"Baby",89,True} 
print(myset,type(myset))
```

```python
# Dictionary {key:value}-->lection which is unordered, changeable and indexed.   
# No duplicate keys.

mydiect={
    "name":"karthick",
    "age":20,
    "course": "python",
    "time":10.30,
    23.89:89,
    90.9:False,
    "place":"chennai"
} 
print(mydiect,type(mydiect))
print(mydiect["age"])
mydiect["age"]=24
print(mydiect[90.9])
print(mydiect["course"])
mydiect["state"]="Tamilnadu"
print(mydiect)         
```

***MUTABLE VS IMMUTABLE OBJECTS***  
**Immutable** ***(int, float, bool, str, tuple are immutable)***  
*Same data stores in Same location  
*Different Data stores in Different location  
**Mutable**  ***((list, set, dict)***  
*Same data stores in Different location  
*Different Data stores in Different location  

```python
# Immutable (int, float, bool, str,tuple are immutable)

a=800 # int
b=800
c=78 # change different value
print(a,type(a),id(a))
print(b,type(b),id(b))
print(c,type(c),id(c))

c=(30,40,50,60,70) # tuple using
d=(30,40,50,60,70)
print(c,type(c),id(c))
print(d,type(d),id(d))
```

```python
# Mutable  ((list, set, dict)

e=[20,40,60,80] #
f=[20,40,60,80]
print(e,type(e),id(e))
print(f,type(f),id(f))
```

***Identity operators***

```python
# Identity operators (is, is not) only based on true or false

a=[10,30,40,50] # using list(mutable)
b=a
c=[10,30,40,50]
print(a==b)  # true        # is and is not discribe location of value .like true or false
print(a is c) # false
print(a is not b) # false

a=(10,30,40,50) # using tuple(mutable)
b=a
c=(10,30,40,50)
print(a==b)  # true     # is and is not discribe location of value .like true or false
print(a is c) # false
print(a is not b) # false
```

***Membership operators***

```python
# Membership operators(True or False)
# in ,not in

colours = ["yellow", "white", "red", "orange"] #read by iteam by iteam

print("white" in colours) #true
print("orange" in colours) # True
                                 # in,ont in is represent in value it availabe or not
print("purple" not in colours) #true
print("yellow" not in colours) # false

x=[56.7,90,56,"python", "class"]
print("python" in x) 
print(90 not in x)
```
