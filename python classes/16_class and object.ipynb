### ***SELF PARAMETER***  
 The self parameter is a reference to the current instance of the class,and is used to access variables that belongs to the class. It does not have to be named self , you can call it whatever you like, but it has to be the first parameter of any function in the class:  
 * note : function inside class should have self as its first argument

```python
# by using def function

class vehicle:
    x=30   #global variable ,it cant change
    y=40

    def myfunc(self):  # The object value is passed to self. self refers to the current object,
            #  When a method is called using an object, that object is automatically passed to self.
        print("hi myfunc is calling")
        print(self)
        
#create object
obj1=vehicle() 
print(obj1.x,obj1.y)

obj1.myfunc()
print(obj1)
```

## **Polymorphism**  
  * The word polymorphism means having many forms. In simple words, we can define polymorphism as the ability of a message to be displayed in more than one form.
   * Real life example of polymorphism, a person at the same time can have different characteristic. Like a man at the same time is a father, a husband, an employee. So the same person posses different behavior in different situations. This is called polymorphism.

```python
print(len("hello baby"))
print(len([10,20,"baby",50]))
```

```python
#function based polymorphism

def add(a,b):
    return a+b
print(add(3,5))
print(add("hi","baby"))
```

```python
#method based polymorphism
class India:
    def captial(self):
        print("India captial is Dehli")
    def language(self):
        print("hindhi is language of India")
     
class USA:
    def captial(self):
            print("USA captial is Washington DC")
    def language(self):
            print("English is language of USA")

obj1_India=India()
obj2_USA=USA()

for country in [obj1_India,obj2_USA]:
     country.captial() # function calling
     country.language()
```

### ***data abstraction***  
 * Data Abstraction - Abstraction means displaying only essential information and hiding the details. Data abstraction refers to providing only essential information about the data to the outside world, hiding the background details or implementation.

```python
class Clothing(object): 
 
 def __init__(self, type, color, size, price=None):
  self.type = type
  self.color = color
  self.size = size
  self.price = price

 def set_price(self, price):
# Set the price of an item of clothing.
  self.price = price
  print(f"Setting the price of the {self.color} {self.type} to ${price}.")
                                                                    
 def get_price(self):
   # Get the price of an item of clothing, if price is set.
   try:
     assert self.price != None
     print(f"The {self.color} {self.type} costs ${self.price}.")
   except:
    print(f"The price of the {self.color} {self.type} hasn't been set yet!")

 def promote(self, percentage):
    # Lower the price, if initial price is set.
    try:
       self.price = self.price * (1-percentage/100)
       print(f"The price of the {self.color} {self.type} has been reduced by {percentage} percent! It now only costs ${self.price:.0f}.")
    except:
        print(f"Oops. Set an initial price first!")

bluejeans = Clothing("jeans", "blue", 12)

bluejeans.get_price()
bluejeans.set_price(100)
bluejeans.get_price()
bluejeans.promote(25)
bluejeans.get_price()
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


class UPI(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


c = CreditCard()
u = UPI()

c.pay(1000)
u.pay(500)
```

### ***inheritance***  
* Inheritance allows us to define a class that inherits all the methods and properties from another class. Parent class is the class being inherited from, also called base class. Child class is the class that inherits from another class, also called derived class.  
 **Single inheritance**:  
  When a child class inherits from only one parent class  
**Multiple inheritance**:  
  When a child class inherits from multiple parent classes                        
**Multilevel inheritance**:   
  When we have child and grandchild relationship.                          
**Hierarchical inheritance**:  
  More than one derived classes are created from a single base.  
**Hybrid inheritance** :  
  Formed by combining the above types of inheritance

```python
#single inheritance

class A:  #parent class
    a=100

    def parentfunc(self):
        print("this is parentfunction is calling")
        print(f"this is A class variable ={self.a}")

class B(A): #child class
    b=200
    def childfunc(self):
     print("this is childfunction is calling")
     print(f"this is B class variable ={self.b}")

obj=B()
obj.parentfunc()
obj.childfunc()

```

```python
# multiple inheritance
class A:  #parent class 1
    a=100

    def parent1func(self):
        print("this is parent1function is calling")
        print(f"this is A class variable ={self.a}")

class B: # parnet class 2
    b=200
    def parent2func(self):
     print("this is parent2function is calling")
     print(f"this is B class variable ={self.b}")

class C: # parnet class 3
    c=300
    def parent3func(self):
     print("this is parent3function is calling")
     print(f"this is C class variable ={self.c}")

class D(A,B,C): # child class 1
    d=400
    def childfunc(self):
     print("this is childfunction is calling")
     print(f"this is D class variable ={self.d}")


obj=D()

obj.parent1func()
print()
obj.parent2func()
print()
obj.parent3func()
print()
obj.childfunc()
print()
```

```python
#multilevel inheritance
class A:     #grand parent class 
    a=100

    def grandfunc(self):
        print("this is grandfunc is calling")
        print(f"this is A class variable ={self.a}")

class B(A): # parnet class 
    b=200
    def parentfunc(self):
     print("this is parentfunc is calling")
     print(f"this is B class variable ={self.b}")

class C(B): # child class 
    c=300
    def childfunc(self):
     print("this is childfunction is calling")
     print(f"this is C class variable ={self.c}")



obj=C()   #wht caaling c ,which means B having A..

obj.grandfunc()
print()
obj.parentfunc()
print()

obj.childfunc()
print()
```

```python
#Hierarchical inheritance:(one parrent -multiple child)

class A:  #parent class 1
    a=100

    def parent1func(self):
        print("this is parent1func calling")
        print(f"this is A class variable ={self.a}")

class B(A): # child class 1
    b=200
    def child1func(self):
     print("this is child1func is calling")
     print(f"this is B class variable ={self.b}")

class C(A): # parnet class 2
    c=300
    def child2func(self):
     print("this is child2func is calling")
     print(f"this is C class variable ={self.c}")

class D(A): # child class 13
    d=400
    def child3func(self):
     print("this is child3func is calling")
     print(f"this is D class variable ={self.d}")

obj1=B()
obj1.parent1func()
print()
obj1.child1func() 

print("-------------------------")
obj1=C()

obj1.parent1func()
print()
obj1.child2func()

print("--------------------------")

obj1=D()
obj1.parent1func()
print()
obj1.child3func()
```

```python
# Hybrid inheritance : Formed by combining the above types of inheritance

class A:
    a=100  #parent class 1

    def myfunc(self):
        print("this is myfunc calling")
        print(f"this is A class variable ={self.a}")

class B(A): 
    b=200
    def newfunc(self):
     print("this is newfunc is calling")
     print(f"this is B class variable ={self.b}")

class C(A): 
    c=300
    def oldfunc(self):
     print("this is oldfuncis calling")
     print(f"this is C class variable ={self.c}")

class D(B,C): 
    d=400
    def yourfunc(self):
     print("this is yourfunc is calling")
     print(f"this is D class variable ={self.d}")

obj=D()
obj.myfunc()
print()
obj.newfunc()
print()
obj.oldfunc()
print()
obj.yourfunc()
print()
```

### **super()**-*function*
* The super() builtin returns a proxy object (temporary object of the superclass) that allows us to access methods of the base class.  
In Python, super() has two major use cases:  
• Allows us to avoid using the base class name explicitly  
• Working with Multiple Inheritance  

```python
#python standard function
class A:
    a=100

    def myfunc(self):
        print("hello A function is calling")
        print("this is A class variable are{self.a}")

class B(A):
    b=200

    def myfunc(self):
        print("hello B function is calling")
        print("this is B class variable are{self.b}")
        super().myfunc()

class C(B):
    c=300

    def myfunc(self):
        print("hello C function is calling")
        print("this is C class variable are{self.c}")
        super().myfunc()

obj=C()
obj.myfunc()
```

### **OPERATOR OVERLOADING MAGIC METHODS**

```python
#__add__(self,other)
class myclass:
    def __init__(self,x):# constructor 
        self.x=x
    def __add__(self,others):
        return f"the addition of {self.x}+{others.x}= " +str(self.x+others.x)

obj1 = myclass(30)
obj2 = myclass(30)
print(obj1 + obj2)  

obj3 = myclass(30)
obj4 = myclass(40)
print(obj3 + obj4)
```

```python
#__mul__(self,other)
class myclass:
    def __init__(self,x):
        self.x=x
    def __mul__(self,others):
        return f"the multiplication of {self.x}*{others.x}= " +str(self.x*others.x)

obj1 = myclass(30)
obj2 = myclass(30)
print(obj1 * obj2)

obj3 = myclass(30)
obj4 = myclass(40)
print(obj3 * obj4)
```
