### **class & object**  
 * A Class is like an object constructor, or a "blueprint" for creating objects.  
 * Objects are instance (copy) of a class that can be created and  
   destroyed later

```python
class vehicle:
    x=30   #global variable ,it cant change
    y=40

#create object
obj1=vehicle() 
obj2=vehicle()

print(obj1.x,obj1.y)

#if you want to change the value in object.
obj1.x="baby"
print(obj1.x,obj1.y)  

print(obj2.x,obj2.y) # display only global vlaue.
```

### ***SELF PARAMETER***  
 The self parameter is a reference to the current instance of the class,and is used to access variables that belongs to the class. It does not have to be named self , you can call it whatever you like, but it has to be the first parameter of any function in the class:  
 * note : function inside class should have self as its first argument

```python
# by using def function

class vehicle:
    x=30   #global variable ,it cant change
    y=40

    def myfunc(self):  # The object value is passed to self. self refers to the current object,
            #  When a method is called using an object, that object is automatically passed to self.
        print("hi myfunc is calling")
        print(self)
        
#create object
obj1=vehicle() 
print(obj1.x,obj1.y)

obj1.myfunc()
print(obj1)
```

```python
class vehicle:
    x=30   #global variable ,it cant change
    y=40

    def myfunc(self,a,b):  # The object value is passed to self. self refers to the current object,
                            #  When a method is called using an object, that object is automatically passed to self  
        print("Hi, myfunc is calling")
        print(f"the function variable of= {a},{b}")
        print(f"the class function are ={self.x},{self.y}")
        
#create object
obj1=vehicle() 
print(obj1.x,obj1.y)


obj1.myfunc(100,300)
print()

obj1.x="baby"
print(obj1.x,obj1.y)
print()

obj1.myfunc(50,79)
print()

obj1.myfunc(100,200)

#here only change the function variable,cant change class variable permentally
```

### ***CONSTRUCTOR - FUNCTION***  
*syntax:*   
**\_\_init\_\_()**


```python
# mostly important one
class xyz:
    def __init__(self,x,y): #constructor  function 
        print("constructor calling")
        self.x=x
        self.y=y

obj1=xyz(30,40)  #we can call mynction dirctly in class,bcz this is construcor
print(obj1.x,obj1.y)
print("--------------------")
obj2=xyz(300,490)
print(obj1.x,obj1.y)
       
```

```python
#real time example

class student:
    def __init__(self,name,age,city):
        print("student data")
        self.name=name
        self.age=age
        self.city=city

student_baby=student("baby",20,"chennai")
print(student_baby.name)
print(student_baby.age)
print(student_baby.city)
print("----------------")
student_karthi=student("karthi",26,"erode")     
print(f"{student_karthi.name}\n{student_karthi.age}\n{student_karthi.city}")

        
```

### ***DESTRUCTORS - FUNCTION***
*syntax:*
**\_\_del_\_\()**

```python
class student:
    def __init__(self,name,age,city):
        print("constructor calling")
        print("student data")
        self.name=name
        self.age=age
        self.city=city

    def __del__(self):
        print("destructors calling")

student_baby=student("baby",20,"chennai")
print(student_baby.name)
print(student_baby.age)
print(student_baby.city)
del student_baby
print("----------------")
student_karthi=student("karthi",26,"erode")     
print(f"{student_karthi.name}\n{student_karthi.age}\n{student_karthi.city}")
print("----------------")
student_priya=student("priya",23,"delhi") 
print(f"{student_priya.name}\n{student_priya.age}\n{student_priya.city}")
print("----------------")
```

## **Polymorphism**  
  * The word polymorphism means having many forms. In simple words, we can define polymorphism as the ability of a message to be displayed in more than one form.
   * Real life example of polymorphism, a person at the same time can have different characteristic. Like a man at the same time is a father, a husband, an employee. So the same person posses different behavior in different situations. This is called polymorphism.

```python
print(len("hello baby"))
print(len([10,20,"baby",50]))
```

```python
#function based polymorphism

def add(a,b):
    return a+b
print(add(3,5))
print(add("hi","baby"))
```

```python
#method based polymorphism
class India:
    def captial(self):
        print("India captial is Dehli")
    def language(self):
        print("hindhi is language of India")
     
class USA:
    def captial(self):
            print("USA captial is Washington DC")
    def language(self):
            print("English is language of USA")

obj1_India=India()
obj2_USA=USA()

for country in [obj1_India,obj2_USA]:
     country.captial() # function calling
     country.language()
```
