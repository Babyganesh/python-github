```python
#palindrome
# 1 Find weather the String is check palindrome

# 1. Input: racecar
    # Output: True
# 2. Input: MalaYa lAm
    # Output: True
# 3 Input: Sir, I demand, I am a maid named Iris
    # Output: True
# 4 Input : Red roses run no risk. Sir, on Nurse's Order.
    # Output: True
```

```python
# 1. Input: racecar
    # Output: True

mystr="racecar"
print(mystr)
revstr=mystr[::-1]
print(revstr)

if mystr==revstr:
    print(True)
else:
    print(False)
```

```python
# 2. Input: MalaYa lAm
    # Output: True

mystr="MalaYa lAm"
print(mystr)
newstr=mystr.lower().replace(" ","")  # lower and replace 

print(newstr)
revstr=newstr[::-1]
print(revstr)

if newstr==revstr:
    print(True)
else:
    print(False)
```

```python
# 3 Input: Sir, I demand, I am a maid named Iris
    # Output: True

mystr="Sir, I demand, I am a maid named Iris"
print(mystr)
newstr=mystr.lower().replace(" ","").replace(',',' ').replace(' ','')

print(newstr)
revstr=newstr[::-1]
print(revstr)

if newstr==revstr:
    print(True)
else:
    print(False)

```

```python
# 4 Input : Red roses run no risk. Sir, on Nurse's Order.
    # Output: True

mystr="Red roses run no risk. Sir, on Nurse's Order."
newstr=''


for each in mystr:
    if each.isalpha():
        newstr=newstr+each.lower()
print(newstr)

if newstr == newstr[::-1]: #reverse
 print(True)
else:
 print(False)
```

```python
def Check_palindrome(mystr):
    newstr=''

    for each in mystr:
     if each.isalpha():
        newstr=newstr+each.lower()
    print(newstr)

    if newstr == newstr[::-1]:
      return True                #return call,reuseable many time
    else:
      return False

ans=Check_palindrome("Mad AM ") #number ignore bcz isalbha function  is used
print(ans)

print(Check_palindrome("racecar") )
print(Check_palindrome("Red roses run no risk. Sir, on Nurse's Order. "))
print(Check_palindrome("Sir, I demand, I am a maid named Iris"))

```

### ***Character frequency***

```python
# input="Hello world"
# output={'h':1,'e':0,'l':3,'o':2,'w':1,'r':1,'d':1}

mystr="Hello world"
newstr=''

for each in mystr:
    if each.isalpha():
        newstr=newstr+each.lower()
print(newstr)
mydict={}

for each1 in newstr:
    if each1 not in mydict.keys():
        mydict[each1]=1   
    else:
        mydict[each1]+=1
print(mydict)
```
