***python file to json file---(json.dumps)***

```python
#json module is default

import json  

my_dict={
    'name':'baby',
    'city':('chenni','erode','cbe'),
    'year':None
}
print((my_dict),type(my_dict))

# change python to json format
my_data=json.dumps(my_dict)
print(my_data,type(my_data))
```

Create a JSON file containing: 
* Name  
* Age  
* Email  
* Skills (tuple)  
* Address (nested dictionary)  
Then:  
* Save it to a file.  
* Read the file.  
* Print only the skills.  
* Add a new skill.  
* Update the file.  
This task covers json.dumps(), json.loads(), json.dump(), and json.load(), which are the most commonly used JSON functions in Python.

```python
# self task (Create Employee Data)
import json

employee={ 
    'Name' :'baby',
    'age'  :'29',
    'email':'babygensanece@gmail.com',
    'Skills':('python','SQL','ml'),
    'address': {
        'city' : 'chennai',
        'state' : 'tamilnadu',
        'pincode': None
        }
}
print("dictionary data")
print(employee,type(employee))
print()

#Convert Dictionary to JSON String (json.dumps)
json_string=json.dumps(employee)
print("json string")
print(json_string,type(json_string))
print()

#Convert JSON String back to Dictionary (json.loads)
python_data=json.loads(json_string)
print("python data")
print(python_data,type(python_data))
print()

#create  and save json file with content
with open("C:\\Users\\Admin\\Desktop\\newdata.json",'w')as my_file:
 json.dump(employee,my_file)

# Read JSON data from file(json.load)
with open(r"C:\\Users\\Admin\\Desktop\\newdata.json", "r") as my_file:
    data = json.load(my_file)

 #Print Only Skills
print("skils")
for each in data["Skills"]:
  print(each)

#Add a New Skill
data['Skills'].append("RAG")

#Update the file WITH NEW SKILLS
with open("C:\\Users\\Admin\\Desktop\\newdata.json",'w')as my_file:
 json.dump(data,my_file)
 print(data)

 print("\nNew skill added and file updated successfully!") 

```
