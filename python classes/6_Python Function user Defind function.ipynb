### **Function with paramater**

```python
def myfunc(): # local varibales link (x.y)
    x = 500
    y = 400
    print("sum  of x & y:",x+y)

myfunc()
```

```python
def myfunc(x,y): # parameter(varibales) - not permenat ,temporay only
   print("x = " ,  x)
   print("y = ", y)
   print("sum  of x & y:",x+y) # not saves in memory location and its local variales

myfunc(20,30) # Arugments 
print()

myfunc(100,300)
print() 
myfunc(500,0)
```

```python
#factorial 

mylist1=[1,2,3,4,5]
mylist2 = [1,2,3,4,5,6]
mylist3 = [1,2,3,4,5,6]

def my_fact(mylist):  
   x = 1    
   for i in mylist:
    x = x*i
   return x

ans=my_fact(mylist1)
print(ans)

print(my_fact(mylist2))
print(my_fact([1,2,3,4,5,6]))
print(my_fact([1,2,3,4,5,6,7,8]))
```
