### ***dictionary function***
keys()--Returns a list containing the dictionary's keys  
values()--Returns a list of all the values in the dictionary  
items()--Returns a list containing a tuple for each key value pair  
pop(key)--Removes the element with the specified key  
popitem()--Removes the last inserted key value pair  
clear()--Removes all the elements from the dictionary  
update()--Updates the dictionary with the specified key value pairs  
get()--Returns the value of the specified key  
copy()--Returns a copy of the dictionary  

```python
mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict,type(mydict))
print(mydict[500])
mydict["City"]='Mumbai'  # city change
mydict["Course"]="Python" # add 
print(mydict)
```

```python
#key()

mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict,type(mydict))
print()

# print(mydict.keys())       #tuple inside list([varialbles])
print(list(mydict.keys()))   # list inside variable[variable] 

ans= list(mydict.keys())
print(ans[2])              # index call(like variable show)

for each in mydict.keys():
    print(each)          # it shows inside variable index like a,g,i---only string[type error]
                         #in for loop ..only print the value
```

```python
# values()
mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict,type(mydict))
print()

print(list(mydict.values()))     # shows value only
print() 

ans=list(mydict.values())
print(ans[3])
print()

for each in mydict.values():
    print(each) 
```

```python
# items()----set of  keys and values

mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict,type(mydict))
print()

print(list(mydict.items()))     # shows keys and value
print() 

for each in mydict.items():  
    # print(each)   
    print(f"{each[0]}\t{each[1]}")             
```

```python
# pop(keys)---dict

mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict)
mydict.pop("Age")  # delete (key: keeper value)
print(mydict)
```

```python
# popitem---dict

mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict)
mydict.popitem() # delete last (key: keeper value)
mydict.popitem() # delete last (key: keeper value)
mydict.popitem() # delete last (key: keeper value)
print(mydict)
```

```python
# clear()---empty dict (delete key and value)

mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict)
mydict.clear()
print(mydict)
```

```python
#update()  ----Updates the dictionary with the specified key value pairs 

mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict)
newdict={"Name":"babyganesh",
             "Age" : 29,
               "City": "erode",
                  500 : 50.55,
                    "class":"python"
        }
newdict.update()
print(newdict)
```

```python
#get()---Returns the value of the specified key 
mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict)
print(mydict[500])
print(mydict[600]) # KeyError: 600
print("Hello Python")
```

```python
#get()---Returns the value of the specified key (None)
mydict = {
           "Name":"baby",
             "Age" : 29,
               "City": "Chennai",
                  500 : 50.55,
         }
print(mydict)
print(mydict.get(500))
print(mydict.get(600)) # None
print("Hello Python")
```

```python
#copy()---copy()--Returns a copy of the dictionary  

x = {"A":100,"B":200,"C":300,}
y = {"A":100,"B":200,"C":300,}

x["d"]=900
y["e"]=800

print(f"X={x},{type(x)},{id(x)}")
print(f"Y={y},{type(y)},{id(y)}")
```

```python
x = {"A":100,"B":200,"C":300,}
y = x.copy()  

x["d"]=900
y["e"]=800

print(f"X={x},{type(x)},{id(x)}")
print(f"Y={y},{type(y)},{id(y)}")
```

### ***Nested Dictionary***

```python
mydict = {
           "Name":"baby",
            "Age" : 29,
            "City": ["Chennai","erode","coimbatore"],
            "year":(2018,2020,2016) 
         }
print(mydict,type(mydict))
print(mydict.keys())
print(mydict.values())
print(mydict.get("Name"))
print(mydict.get("City"))
print(mydict.get("City")[0])  #print chennai
print(mydict.get("City")[2])  #print coimbatore
print("-----------------------")
print(mydict.get("place"))   # none
print(mydict.get("year")[1])   # 2020
print("-----------------------")
print(mydict.get("City")[2][3:])  #in city -2nd index --starting from 3 to till end

```

```python
x = {
    "menu": {
             "id": "file",
             "value": "File",
             "popup": {
                        "menuitem": [
                           {"value": "New", "onclick": "CreateNewDoc()"},
                           {"value": "Open", "onclick": "OpenDoc()"},
                           {"value": "Close", "onclick": "CloseDoc()"}
                                     ]
                      }
            }
    } 
print(x,type(x))
print(x.keys())
print(x.get("menu"))
print(x.get("menu")["popup"])
print(x.get("menu")["popup"]["menuitem"])
print(x.get("menu")["popup"]["menuitem"][0])
print(x.get("menu")["popup"]["menuitem"][1])
print(x.get("menu")["popup"]["menuitem"][2])
print(x.get("menu")["popup"]["menuitem"][0]["onclick"])
print(x.get("menu")["popup"]["menuitem"][2]["onclick"])
print(x.get("menu")["popup"]["menuitem"][1]["value"])


```
