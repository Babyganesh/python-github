## **RegEx Functions**  
   import re
* findall()-Returns a list containing all matches  
* search() - Returns a Match object if there is a match anywhere in the string  
* split() - Returns a list where the string has been split at each match  
* sub() - Replaces one or many matches with a string  

### **findall()-Returns a list containing all matches**

```python
import re

with open("C:\\Users\\Admin\\Desktop\\baby2.txt") as myfile:
    mydata=myfile.read()

    list=re.findall(r"[Aa][n][d]",mydata)
    print(list)
    
```

```python
#date format
import re

with open("C:\\Users\\Admin\\Desktop\\baby2.txt") as myfile:
    mydata=myfile.read()

    list=re.findall(r"\d{2}[-/.]\d{2}[-/.]\d{4}",mydata)
    print(list)
```

## **split()**
*  import re
* .spilt()-list['abc','abc']

```python
import re
with open("C:\\Users\\Admin\\Desktop\\baby2.txt") as myfile:
    mydata=myfile.read()

    list=re.split(r" ",mydata) #cut the content ,store in list
    print(list)

    for each in list:
        print(each)
```

## **search()**
Returns a Match object if there is a match anywhere in the string  
 * search first occurance only   
 * "and"--->print 1st and(ex)


```python
import re

with open("C:\\Users\\Admin\\Desktop\\baby2.txt") as myfile:
    mydata=myfile.read()

    list=re.search(r"[a][n][d]",mydata) # take first occurance value
    print(list)
```
