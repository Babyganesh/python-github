### **FILE HANDLING**  
 File handling is an important part of any web application. Python has several functions for creating,    reading, updating, and deleting files.  
   * open()function  
     •The key function for working with files in Python is the open() function.  
     •The open() function takes two parameters; filename, and mode.    
     •There are four different methods (modes) for opening a file:  

```python
myfile=open("C:/Users/Admin/Desktop/baby.txt")

#  check curser is where it located.
print(myfile.tell())

#read the all content  in file and store in data variable
mydata=myfile.read(100)
print(mydata)

# curser palced at last 
print(myfile.tell())

# agin read the content
mydata=myfile.read(200)
print(mydata)
print(myfile.tell())   # ans = 300

#for return to 0 position 
myfile.seek(0)
print(myfile.tell())

mydata=myfile.read(500)
print(mydata)
print(myfile.tell())

#close the file
myfile.close()
```

## **with keyword**
alternative method

we have 3 mode  
1.read()---'r'  
2.append()---'a'  
3.write() ---'w'
4.create()---'x' 

```python
# 1.read()  

# here used with keyword..only use open().
# if we not have copy as path in file(hold shift,press right clike )
with open("C:\\Users\\Admin\\Desktop\\baby.txt",'r') as myfile:
   
#    myline=myfile.readline() # readline -print 1st line
#    print(mydata)

   mylines=myfile.readlines() # readlines -print all line in list at onle line
   print(mylines)
   print("-------------------")

   print(mylines[2]) #print 1 line
# #    print(mydlines[1]) #print 2 line

   for each in mylines:
      print(each)
  
```

## **2.append()---'a'**
* write ()   
* cursor palced at end of the  line  
* if we need to add new line content .can use \n  
  by using with keyword \--->\\  

```python
#2.append()---'a' 

with open("C:\\Users\\Admin\\Desktop\\baby.txt", 'a') as myfile:
    print(myfile.tell())

    mydata=myfile.write("\nhello world")
 
```

## **3.write() ---'w'**
 * also called as edit mode
 * already exiting file deleted -dont use write mode 
 * but add new content-->using write("content")
 * if we want to edit the content ,can use normal change in content
    like hello world-->hello my world


```python
#write mode
with open("C:\\Users\\Admin\\Desktop\\baby.txt", 'w') as myfile: # dele all content if we use 'w'mode
    print(myfile.tell())

    # add content ,using write()
    mydata=myfile.write("hi this is baby ganesh!!!!!!!") #if need change automative correct here
```
