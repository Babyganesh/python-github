### **Decorator**  
To understand decorator we need to understand four concepts  
• nested function  
• function as a reference  
• function can return another function  
• function as a parameter  

```python
def outer():
    n1=10
    def inner():
        n2=40
        n3=n1+n2
        return n3
    return inner
ans=outer()    # here ,pass the inner function in outer function
print(ans())   #here ans is one of the function so,we can call   ans the  function 
print(ans.__name__)
```

```python
#function can return another function
def create_adder(x):
    def adder(y):
       return x+y 
       
    return adder
   
ans = create_adder(100)
print(ans(50))         #here 50 is y value.       
print(ans.__name__)
```

```python
#this is written without decorators
# without distrubing this function we are going to add some functionality to it,

def strupper(func):

   def inner():         #to modify existing functions functionality
     str1 = func()      #strname() is called and str="hi welcome "
     return str1.upper()
   return inner

@strupper                  #another method @ strupper (outerfunction name)
def strname():
   return "hi world"

@strupper
def myname():
   return "BaBY"

print(strname())
print(myname())
```

```python
#decorator with parameter
#existing function
def div(a,b):
 return a/b

print(div(10,2))
print(div(10,5))
print(div(10,0))
```

```python
#adding decorators
def divdec(func):
    def inner(x,y):
        if y==0:
            return "dont givr divisor as zero"
        else:
            return func(x,y)
    return inner #as it returns inner, inner() will be called (function passed as refernce)

@divdec
def div(a,b):
    return a/b

print(div(20,3))
print(div(30,0))  #here occure zerodivisionerror,if you aviod the erroe by using @divedec func name
print(div(30,3))
```

### **Types of methods**  

• instance method  
• class method  
• static method  


```python
class Subject:
    def __init__(self,tamil,english,maths):
        self.tamil=tamil
        self.english=english
        self.maths=maths
       
    def marks(self):    # Instance method 
         Total = self.tamil + self.english + self.maths
         return Total


Kumar = Subject(90, 75, 100)
Mani = Subject(70, 50, 60)
Kavi = Subject(40, 35, 20)

result1 = Kumar.marks()
result2 = Mani.marks()

print(result1)
print(result2)

print(Kavi.marks())
```

• class method  
• parameter must be cls  
• should use classmethod decorator  

```python
class demo:
   
   def method(self,a,b):
    return "The Total is "+ str(a + b)
   
   
   def method1(cls): #cls is a naming convention (not a rule ) #cls=class name here return 
                          #"Hi im a class decorator"
    return "Hi im a class decorator"

obj=demo()
print(obj.method(30,20) )
print(obj.method1())

print("-------------------")

print(print(obj.method(30,10) ))

print(print(obj.method1()))
```

```python
#Using @classmethod, we can access the method directly from the class 
# without creating an object.

class demo:
   @classmethod
   def method(self,a,b):
    return "The Total is "+ str(a + b)
   
   @classmethod
   def method1(cls): #cls is a naming convention (not a rule ) #cls=class name here return 
                          #"Hi im a class decorator"
    return "Hi im a class decorator"

# obj=demo()
# print(obj.method(30,20) )
# print(obj.method1())

print("-------------------")

print(obj.method(30,10) )

print(obj.method1())
      
```

```python
class demo :
  x = "hi i a class method"

  @classmethod
  def method(self,a,b):
     return "The total is "+str(a+b)
    
  @classmethod
  def method1(cls,x,y): #cls is a naming convention (not a rule ) #cls=class name
     print("The total is "+str(x*y))
     return cls.x  # return the class variable X

obj = demo()

print(obj.method(50,60))
print(obj.method1(30,2))
print('--------------')
print(demo.method1(70,3)) # by using @classmethod-dirctly call (class name.func name)
print(demo.method(70,3))
```

### **static method**
* no self or cls parameters
* use decorator @static method


```python
# (A static method belongs to the class but does not use self (object) or # cls (class).

class demo :
   var="Hi im a class method"

   def method(self,a,b):
        return "The total is "+str(a+b)

   @staticmethod
   def method1(cls,x,y):
        print("the multiplication of " +str(x*y))
        return demo.var   

   @staticmethod
   #  (A static method belongs to the class but does not use self (object) or # cls (class).                             
   def method2():
        return " hi this is static method"

obj= demo()

print(obj.method(50,40))
print(obj.method1(50,4))
print(obj.method2())  
```
