### **Scope**
Types of scope  
* **local scope**  
    * Inside a specific function function  
    * Accessible only within the function where it was created
* **Global scope**  
    * Outside of any function (at the top level of the script)  
    * Accessible anywhere in the program (inside and outside functions

```python
x = 10
y = 20 

print(f"X in Global scope = {x}")
print(f"Y in Global scope = {y}")
print() 

def myfunc(x=0,y=0):
  print(f"X in Local scope = {x}")
  print(f"Y in Local scope = {y}")

myfunc()
print()

myfunc(1000,500)
myfunc()
```

## **Recursion**  
reverse

```python
def myadd(x,y):
    return x + y
def mysub(x,y):
    return x - y
#        15        30          15
print(mysub(myadd(10,20),myadd(5,10)))
```

```python
 # factorial   

def myfactorial(num):
    pro = 1
    for i in range(1,num+1):
      pro = pro * i
    return pro

ans = myfactorial(5)
print(ans)
```

```python
def myrec_factorial(num):
    
    if num == 1:
     return 1
    else:
     return num * myrec_factorial(num - 1)
    
ans = myrec_factorial(5)
print(ans)
```
