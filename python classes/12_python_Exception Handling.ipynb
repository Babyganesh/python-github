### **Handling error with raise, Except & as**

```python
while True:
    try:
        num = int(input("Enter a number inbetween 10-50:"))
        if num<10:
            raise ValueError("Please Enter a number greater than or equal to 10")
        elif num>50:
            raise ZeroDivisionError("Please Enter a number lesser than or equal to 50")
        else:
            print(f"Thank you for entering the = {num}")
            break
     
    except ValueError as x:
        print(x)
    except ZeroDivisionError as y:
        print(y)
```

### **What is Nested Error Handling in Python?**

```python
#nested loop
try:
    print("Outer try block")

    try:
        num = int(input("Enter a number: "))
        result = 10 / num
        print(result)

    except ValueError:
        print("Inner Error: Invalid input (not a number)")
    except ZeroDivisionError:
        print("Inner Error: Cannot divide by zero")

except Exception as e:
    print("Outer Error:", e)
```

## **Task 1: List Index and Input Error** 
Create a program that:  
• Gets a list of numbers from the user.  
• Asks the user to enter an index.  
• Use nested try-except blocks to handle:  
• ValueError  
• IndexError   

```python

numbers = [10, 20, 30, 40, 50]
print("Your list of numbers is:", numbers)


try:
    # Ask the user to enter an index
    user_input = input("Enter an index to look up: ")
    
    try:
        #  convert the input into an integer
        index = int(user_input)  #  raise ValueError
        
        #  access the list item at that index
        result = numbers[index]  #  raise IndexError
        
        print(f"The number at index {index} is {result}.")
        
    except IndexError:
        #   index is out of range
        print("Error: That index does not exist in the list!")
        
except ValueError:
    #    entered  instead of a number
    print("Error: You must enter a valid whole number!")
```

```python
try:
    n = int(input("enter int number: "))  # LIKE n=5
    mylist = [] # n is convert to list [0......5]
    print(list(mylist))

    for i in range(n): # range(5)
        num = int(input("Enter number: ")) # enter 5 times like 10,20,30,40,50    
        mylist.append(num) #adds an item to the end of the list 
                            # [10,20,30,40,50]
        print(list(mylist))

    try:
        index = int(input("Enter an index: ")) # like 3  
        print(f"Value at index {index}: {mylist[index]}") #value=mylist[3]=40

    except ValueError: #entered  instead of a number
        print("Enter a valid index.")

except IndexError: #entered  out of range index 
    print("Index is out of range.")
```

student = {

Ask the user for a key and display its value.  
Use nested exception handling to handle:  
• Invalid input  
• KeyError  

```python
#task 2

student = {
    "name": "baby",
    "age": 29,
    "course": "Python"
}

try:
    key = input("Enter a key (name, age, course): ")

    try:
        print("Value:", student[key])

    except KeyError as k:
        print(" Key not found in the dict.",k)

except ValueError as v:
    print(" Invalid input.",v)
```
