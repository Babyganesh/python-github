***python datatypes***

```python
# Integer-- 1, 1244, 1234 
# Float-- 1.0, 1e10, 123.324
# complex --1+0j (j represents that its complex)
# string ––“Hello world”, “a”, ‘
# boolean --True, False

Student_name="priya" 
height=5.8
weight=5e3
id=2+89j
Age=25
is_student= True

print(Student_name,type(Student_name))
print(height,type(height))
print(weight,type(weight))
print(id,type(id))
print(Age,type(Age))
print(is_student,type(is_student))
```

***Trio:value,id,type***

***Type Casting***(most important -python standed functions)
***•int(),•float(),•str(),•bool()***

```python
# Numerical values can be converted to int, float, bool, or str, 
# but a character cannot be converted directly to float,ini.

# a="100"
# a2=float(a)
# a3=str(a)
# a4=bool(a)
# print(a1)
# print(a2)
# print(a3)
# print(a4)

# int() - converts a value to integer
a = "100"
b = int(a)
print("int():", b, type(b))

# float() - converts a value to floating-point number
c = "123.09"
d = float(c)
print("float():", d, type(d))

# str() - converts a value to string
e = 23
f = str(e)
print("str():", f, type(f))

# bool() - converts a value to Boolean
g = 67
h = bool(g)
print("bool():", h, type(h))
```
