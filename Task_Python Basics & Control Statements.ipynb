```python
# 30.	Write a program using while loop to calculate the sum of numbers from 1 to 100.

total_sum =0
current_number = 1

# Execute the loop until current_number exceeds 100
while current_number <= 100:
    total_sum += current_number  # Add the current number to the running total
    current_number += 1          # Increment the number by 1 for the next iteration

print("The sum of numbers from 1 to 100 is:", total_sum)
```

```python
# 32.	Write a program using while loop to find the factorial of a number

n =int(input("enter the number"))
fact = 1
x=1
while x<=n:
    fact=fact*x      # 1x1=1
                     # 1x2=2
                     # 2x3=6
                     # 6x4=24
                     # 24x5=120......
    x=x+1     
print("Factorial =", fact)  

```

```python
# 41.	Write a program using for loop to find the factorial of a number.
# Formula: \(n! = n * (n - 1) * (n - 2) *(n-3) ......)
# like 4!=4*(4-1)*(4-2)*(4-3)
# 4!=4*3*2*1
# 4!=24

n =int(input("enter the number"))
fact = 1

for i in range(1,n+1):
    fact =fact*i   # 1x1=1
                   # 1x2=2
                   # 2x3=6
                   # 6x4=24
                   # 24x5=120....

print("Factorial =", fact)    
```
