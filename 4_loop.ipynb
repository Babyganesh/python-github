### ***While loop (conditional statement)***  
*A while loop in Python is used to repeat a block of code as long as a condition is True  
* while  
* break  
* continue  
###  ***syntax:***
***while<\condition\>:***

```python
# while loop
# syntax:
# while<\condition\>:

#print 100 to 0 in order
a = 1                        #initilization / start
while a<=20:                # condition/end
    print(a, end = " ")      # if we need horizontal, we can use <end=" ">
    a += 1                  #incrementational/jump
print("\n loop over")        # for new line ---><\n"
```

```python
# print 100 to 0 in reverse oreder
a = 100                   
while a >= 0:             
    print(a, end = " ")
    a -= 10               
print("\n loop over")
```

```python
# Goal: Print numbers from 5 down to 1, and then print "Blast off!".
b=5
while b>0:
    print(b)
    b-=1
print("Blast off!")
```

```python
# break (stop the loop)
 # infinite loop
i = 1
while i <= 10:    
    if i == 5:
        break
    print(i)          
    i += 1             
```

```python
#  Stop at a Specific Number
""" Task: Write a program using while True that starts a counter at 1. 
Print the number and increase it by 1 each time. Use a break statement to stop the l
oop when the counter reaches 5.
Hint: Check if counter == 5 inside an if statement, then call """

count=1
while True:
    print(count)
    count+=1
     
    if count==5:
        break    
```

```python
# Password Checker
""" Task: Create a loop that asks the user to type a password using input(). 
If they type "secret", print "Access Granted" and use break to exit the loop. If they type anything else,
print "Try again".Hint: Combine while True: with an if/else block """

while True:
    a=input("enter the password")
   
    if a=="secret":
     print("Access Granted")  
     break 
    else:
       print("try agin")
```

### ***continue***  
*go back to the top of my loop

```python
# continue 
i = 0

while i < 5:
    i += 1

    if i == 3:
        continue

    print(i)     
```

#### ***Nested Whlie Loop***

```python
# while loop
""" print 12345
      12345
      12345
      12345 """
a=1
while a<=4:
    b = 1
    while b<=5:        
        print(b,end="")
        b += 1
    a += 1 
    print()
```

```python
#  task
#  1. Take row and column inputs from the user
# The input() function reads data as a string, so we convert it to an integer usin
row=int(input("enter a row number"))
cloumn=int(input("enter a cloumn number"))
i = 1
while i <= row:
    j = 1
    while j <= cloumn:
        print(j, end="")
        j += 1
    print()
    i += 1
```

### ***Python for loop***
syntax  

***for<variable_name> in <str/list/tuple/set/diec>:***

```python
# for loop (for<variable_name> in <str/list/tuple/set/diec>)
# using stri
mystr="Hi python"
for a in mystr:      # Fetches Letter by letter
    print(a,end="")
```

```python
# using list
mylist=[34,78,90,60.8,"hello",89,True]
for b in mylist:
     print(b,end=",") 
```

```python
# using tuple
mytuple=(23,56,78,False,"python")
for c in mytuple:
    print(c)
```

```python
# using set
myset={"baby",29,"chennai"}
for d in myset:
    print(d)
```

```python
# using diect
# mydiec={"name":"baby","age":29,"place":"chennai"}
# for e in mydiec:
#     print(mydiec[e])

mydiec = {"name": "baby", "age": 29, "place": "chennai"}  # by using keyand value together
for key in mydiec:
    print(key, mydiec[key])
    
```

### ***Nested for loop***

```python
""" print 1234
        1234
        1234
        1234 """
for i in range(4):
    for y in range(1,4+1):
        print(y,end='')
    print()
```
