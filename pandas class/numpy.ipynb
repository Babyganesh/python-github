```python
import numpy as np
```

```python
#1D array

array_1=np.array([10,20,30])

print(array_1)
print(array_1.shape)
print(array_1.size)
array_1.ndim  # to find what dimentional
array_1.dtype # to find what data type
```

```python
#2D array

array_2 = np.array([[10,20,30],
                    [40,50,60],
                    [70,80,90]])
print(array_2)
print(array_2.shape)
print(array_2.size)
array_2.ndim
array_2.dtype
```

```python
#3D array

array_3 = np.array([[[10,20,30],[10,20,30],[10,20,30]],
                    [[40,50,60],[40,50,60],[40,50,60]]])
print(array_3)
print(array_3.shape)
array_3.size
array_3.ndim
array_3.dtype
```

```python
# type casting
```

```python
array_2 = np.array([[10,20,30],
                    [40,50,60],
                    [70,"hai",90]])
print(array_2) 
print(array_2.dtype) 

#<U21 means under 21 bits
```

```python
# to change ini value 

array_2 = np.array([[10,20,30],
                    [40,50,60],
                    [70,"hai",90]],dtype="int")
print(array_2) 
print(array_2.dtype) 
```

```python
array_2 = np.array([[10,20,30],
                    [40,50,60],
                    [70,"100",90]],dtype="int")
print(array_2) 
print(array_2.dtype) 
```

```python

#to change float
array_2 = np.array([[10,20,30],
                    [40,50,60],
                    [70,"100",90]],dtype="float")
print(array_2) 
print(array_2.dtype) 
```

### **Array filling --(.full)**

```python
# a=np.full((5,5),2)

# or
# a=np.full((5,5),[1,2,3,4,5])
a=np.full((5,5),['B','A','B','Y','G'])
print(a)
print(a.dtype)
```

```python
b=np.full((2,3,4),4)  
print(b)
print(b.dtype)
```

```python
c=np.zeros((5,5),dtype="int") 
print(c)
```

```python
d=np.ones((5,5),dtype="int") 
print(d)
```
