### **dataframe**

```python
# # DataFrame -> 2D Data

Car_Name=pd.Series(["TATA","HUNDAI","BENZ","AUDI","BMW"])
Car_Name

Car_Colour=pd.Series(["GRAY","RED","BLACK","BLUE"])
Car_Colour

# AT SAME TIME WE CANT RUN 2 VARIBLE SO WNAT TO CRAETE ANOTHER VARIABLE NAME

Car_Data=pd.DataFrame({
    "NAME": Car_Name,
    "COLOUR": Car_Colour
})
print(Car_Data)  # if ,not added value it show NaN(not a number)
```

```python
print(Car_Data.dtypes)
print(Car_Data.shape) # how many rows and column
print(Car_Data.size) # row*column
```

```python
# info function
Car_Data.info() #information about data
```

```python
# isna means isnull value is present ?
Car_Data.isna()   #here -mention not value present means is true
```

```python
#show all null value we can use sum
Car_Data.isna() .sum()   
```
